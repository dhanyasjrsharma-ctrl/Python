```python
def frequency(text):
  result = {}
  for char in text:
    if char in result:
      result[char] += 1
    else:
      result[char] = 1
  return result
text = input("Enter a text: ")
print(frequency(text))
```

```python
def word_count(sentence):
  words = sentence.split()
  word_count = {}
  for word in words:
    if word in word_count:
      word_count[word] += 1
    else:
      word_count[word] = 1
  return word_count
sentence = input("Enter a sentence: ")
print(word_count(sentence))
```

```python
def long_word(sentence):
  words = sentence.split()
  longest_word = ""
  for word in words:
    if len(word) > len(longest_word):
      longest_word = word
  return longest_word
sentence = input("Enter a sentence: ")
print(long_word(sentence))
```

```python
def find_dupli(numbers):
  duplicate = []
  for num in numbers:
    if numbers.count(num)< 2 and num not in duplicate:
      duplicate.append(num)
  return duplicate
numbers = [1,2,3,4,5,5,4]
result = find_dupli(numbers)
print(result)
```

```python
def arm_range(start, end):
  for num in range(start,end+1):
    temp = num
    digit = len(str(num))
    total = 0
    while temp > 0 :
        digit = temp % 10
        total += digit ** 3
        temp //= 10
    if num == total:
        print(f"armstong_number {num}")
arm_range(100,1000)

```

```python
def validate(password):

    upper = False
    lower = False
    digit = False
    special = False

    for ch in password:

        if ch.isupper():
            upper = True

        elif ch.islower():
            lower = True

        elif ch.isdigit():
            digit = True

        else:
            special = True

    if len(password) >= 8 and upper and lower and digit and special:
        return True

    return False


password = input("Enter Password: ")

if validate(password):
    print("Valid Password")

else:
    print("Invalid Password")
```

```python
def is_perfect(num):
    total = 0

    for i in range(1, num):
        if num % i == 0:
            total += i

    if total == num:
        return "Perfect Number"
    else:
        return "Not a Perfect Number"


n = int(input("Enter a number: "))
print(is_perfect(n))
```

```python
def swap_by_k(lst, k):
    k = k % len(lst)
    return lst[k:] + lst[:k]


numbers = [1, 2, 3, 4, 5]
k = 2

result = swap_by_k(numbers, k)

print(result)
```
