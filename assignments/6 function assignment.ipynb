## Question 7: Create a function that returns the sum of numbers from 1 to N.


```python
# Input: 10

# Output: 55

def sum():
        n = int(input("enter your number:-"))
        sum = 0
        for i in range(1,n+1):
                sum += i
        return sum 
print(sum())
```

## Question 9: Create a function that returns the factorial of a given number.


```python
# Input: 5

# Output: 120

def fact():
    a = int(input("enter your number:-"))
    fact=1
    for i in range(1,a+1):
        fact*=i
    return fact
print(fact())
```

## Question 10: Create a function that checks whether a number is Prime or Not Prime.


```python
# Input: 13

# Output: Prime Number

def prime():
    n=int(input('enter your number:-'))
    for i in range (2,n):
        if n % i == 0:
            print ("not prime")
            break
    else :
        print ("prime")
prime()
    
```

## Question 13: Create a function that checks whether a string is a palindrome.


```python
# Input: madam

# Output: Palindrome

def palindrome():
    name = input("enter your string:-")
    reverse = name[::-1]
    if reverse == name:
        print(f"{name} is palindrome")
    else:
        print(f"{name} is not palindrome")
palindrome()
```
