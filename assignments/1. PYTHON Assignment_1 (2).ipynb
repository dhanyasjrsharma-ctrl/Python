Q2. Check Data Types using type()?

    INPUT-
    name = "YOUR NAME"
    number = 100
    decimal = 10.5
    status = False

    OUTPUT-
    <class 'str'>
    <class 'int'>
    <class 'float'>
    <class 'bool'>

```python
name=input("enter your name:-")
age=int(input("enter your age:-"))
height=float(input("enter your height:-"))
student=bool(input("are you a student?"))

print(type(name))
print(type(age))
print(type(height))
print(type(student))
```
