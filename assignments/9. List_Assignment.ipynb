### Q4. Count Even Numbers
```python
nums = [12, 5, 8, 7, 10, 13, 16]
```
Count how many even numbers are present.

**Expected Output**
```python
Total Even Numbers = 4
```


````python
nums = [12,5,8,7,10,13,16]
count = 0
for i in nums :
    if i % 2 == 0:
        count += 1
print(count)
````

### Q5. Find Sum of List

### Q6. Find Largest Number

nums = [23, 45, 12, 89, 34]

### Q8. Create a New List of Squares
```python
nums = [1, 2, 3, 4, 5]
```
Create a new list containing squares of all numbers.

**Expected Output**
```python
[1, 4, 9, 16, 25]
```


````python
nums = [1,2,3,4,5]
squared_list = []
for i in range(len(nums)):
    squared_list.append(nums[i]**2)
print(squared_list)

````

### Q9. Count Positive and Negative Numbers
```python
nums = [10, -5, 20, -8, 15, -2]
```
Count positive and negative numbers separately.

**Expected Output**
```python
Positive = 3
Negative = 3
```


````python
nums = [10,-5,20,-8,15,-2]
neg = []
pos = []
pos_count = 0
neg_count = 0
for i in nums :
    if i < 0:
        neg.append(i)
        neg_count += 1
    else:
        pos.append(i)
        pos_count += 1
print("positive = ",pos_count)
print("negative = ",neg_count)
````
