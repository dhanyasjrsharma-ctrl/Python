Q2. Write a Python program to print multiplication table of a number entered by the user.

               5 x 2 = 10
               5 x 3 = 15
               .
               .
               .
               .
               .
               5 x 10 = 50

```python
n=int(input("enter your num:"))
for i in range(n,n*10+1,n):
    print(i)
```

Q4. Write a Python program to find factorial of a number.

    INPUT - Enter a number: 5
    
    OUTPUT - Factorial = 120

```python
n=int(input("enter a number"))
fact=1
for i in range(1,n+1):
    fact=fact*i
print(fact)
```

Q7. Write a Python program to find largest number among 5 entered numbers.

    INPUT -   Enter number: 10
              Enter number: 25
              Enter number: 5
              Enter number: 40
              Enter number: 12
              
    OUTPUT -  Largest number = 40

```python
n1=int(input("enter a num"))
n2=int(input("enter a num"))
n3=int(input("enter a num"))
n4=int(input("enter a num"))
n5=int(input("enter a num"))
l=[n1,n2,n3,n4,n5]
max=l[0]
for i in l:
    if i>max:
        max=i
print(max)
```
