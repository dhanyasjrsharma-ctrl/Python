## Q12. Mini Task

```
Input:
s1={10,20,30}
s2={20,30,40}

Tasks:
1. Union
2. Intersection
3. Difference
4. Symmetric Difference
```

````python
s1 = {10,20,30}
s2 = {20,30,40}
print(s1.union(s2)) # dono sets ke sare elements
print(s1.intersection(s2)) # only common elements in both sets
print(s1.difference(s2)) # s1 me h s2 me nhi h vo elements
print(s1.symmetric_difference(s2)) # common wale elements chhod ke baki sab as output
````
