## Q2. Create an empty set and add numbers 5, 10, 15, 20, 25 using a loop. Print the set.

```
{5, 10, 15, 20, 25}
```

````python
#whithout using a loop
# s = set()
# s.update([5,10,15,20,25])
# print(s)

# using a loop
s = set()
for i in range (5,26,5):
    s.add(i)
print(s)
````
