Q4. Write a Python program to take a string from the user and perform slicing.
    Print:
    - First 5 characters
    - Characters from index 2 to 7
    - Last 4 characters
    
    INPUT - Programming
    
    OUTPUT - 
    
    First 5 characters: Progr
    Characters from index 2 to 7: ogramm
    Last 4 characters: ming
    

```python
word=input("enter your word")
print(f"first five characters: {word[0:5:1]}")
print(f"characters from index 2-7: {word[2:8:1]}")
print(f"last four characters:{word[7:13:1]}")
```
