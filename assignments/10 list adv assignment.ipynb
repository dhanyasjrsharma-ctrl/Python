## Q1. Sort List in Ascending Order

Input
nums = [45, 12, 78, 23, 5]

Output
[5, 12, 23, 45, 78]

```python
nums = [45,12,78,23,5]
for i in range(len(nums)):
    for j in range (i+1,(len(nums))):
        if nums[i]>nums[j]:
            nums[i],nums[j]=nums[j],nums[i]
print(nums)
```

## Q2. Sort List in Descending Order

Input
nums = [45, 12, 78, 23, 5]

Output
[78, 45, 23, 12, 5]

```python
nums = [45,12,78,23,5]
for i in range(len(nums)):
    for j in range (i+i,(len(nums))):
        if nums[i]<nums[j]:
            nums[i],nums[j] = nums[j],nums[i]
print(nums)
```

## Q3. Find Largest Number

## Q5. Remove Duplicate Elements

Input
nums = [1, 2, 3, 2, 4, 1, 5]

Output
[1, 2, 3, 4, 5]

```python

nums = [1,2,3,2,4,1,5]
count = []
for i in nums:
    if i not in count:
        count.append(i)
print(count)
```

## Q10. Find Sum of Elements at Even Index

Input
nums = [10, 20, 30, 40, 50]

Output
90

```python
nums = [10,20,30,40,50]
sum = 0
for i in range(len(nums)):
    if i % 2 == 0:
        sum += nums[i]
print(sum)
```
