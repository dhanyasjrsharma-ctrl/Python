
# Q1. Addition Function

## Problem Statement
Create a function that takes two numbers and returns their sum.

## Sample Input
```python
10, 20
```

## Sample Output
```python
30
```


````python
def add():
    n1=int(input("enter your number"))
    n2=int(input("enter your number"))
    c=n1+n2
    return c
print(add())
    
````


# Q7. Average Function

## Problem Statement
Create a function that takes marks of 3 subjects and returns the average.

## Sample Input
```python
70, 80, 90
```

## Sample Output
```python
80.0
```


````python
def avg():
    n1 = int(input("enter the marks of frist subject:-"))
    n2 = int(input("enter the marks of second subject:-"))
    n3 = int(input("enter the marks of third subject:-"))
    c = (n1 + n2 + n3) / 3
    return c
print(avg())
````


# Q8. Maximum Number Function

## Problem Statement
Create a function that takes two numbers and returns the greater number.

## Sample Input
```python
25, 40
```

## Sample Output
```python
40
```


````python
def greater():
    n1=int(input("enter first number"))
    n2=int(input("enter second number"))
    if n1 > n2 :
        return n1 
    else :
        return n2
print(greater())
````


# Q10. Full Name Function

## Problem Statement
Create a function that takes first name and last name and returns the full name.

## Sample Input
```python
Rahul, Sharma
```

## Sample Output
```python
Rahul Sharma
```


````python
def full_name():
    a = input("enter your first name:- ")
    b = input("enter your second name:-")
    c = a + b
    return c
print(full_name())
````
